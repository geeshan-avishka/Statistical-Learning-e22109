```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Comparing Global Error and Optimal Error

If the bank ignores the risk class \(Y\) and uses the global mean

$$
\mathbb{E}[X]=0.30
$$

as its estimate, the resulting error is the **total variance of \(X\)**.

## Calculating the Total MSE

When the estimate is a constant

$$
c=\mathbb{E}[X],
$$

the Mean Square Error is simply the variance of \(X\):

$$
\text{MSE}_{\text{global}}
=
\mathbb{E}\left[(X-\mathbb{E}[X])^2\right]
=
\mathrm{Var}(X).
$$

Since \(X\) is a Bernoulli random variable with

$$
P(X=1)=0.30,
$$

its variance is

$$
\mathrm{Var}(X)
=
p(1-p)
=
0.30(0.70)
=
0.21.
$$

Therefore,

$$
\text{MSE}_{\text{global}}=0.21.
$$

# Comparing the Errors

## Optimal Error (using \(Y\))

$$
\text{MSE}_{\text{min}}
=
\mathbb{E}[\mathrm{Var}(X\mid Y)]
=
\frac{11}{60}
\approx 0.1833.
$$

## Global Error (ignoring \(Y\))

$$
\mathrm{Var}(X)
=
0.21.
$$

By using the information contained in \(Y\), the bank reduces its prediction error from

$$
0.21
\quad \text{to} \quad
0.1833.
$$

# Law of Total Variance

The **Law of Total Variance** states that

$$
\mathrm{Var}(X)
=
\mathbb{E}[\mathrm{Var}(X\mid Y)]
+
\mathrm{Var}(\mathbb{E}[X\mid Y]).
$$

This decomposes the total uncertainty into two parts.

## Unexplained Variance

$$
\mathbb{E}[\mathrm{Var}(X\mid Y)]
=
0.1833.
$$

This is the remaining uncertainty after using the best possible predictor based on \(Y\).

## Explained Variance

The second term measures how much uncertainty is removed by using \(Y\):

$$
\mathrm{Var}(\mathbb{E}[X\mid Y]).
$$

Since

$$
\mathbb{E}[X\mid Y]
=
\begin{cases}
\frac{1}{6}, & \text{with probability } 0.6,\\
\frac{1}{2}, & \text{with probability } 0.4,
\end{cases}
$$

we compute

$$
\mathrm{Var}(\mathbb{E}[X\mid Y])
=
\mathbb{E}\left[(\mathbb{E}[X\mid Y])^2\right]
-
(\mathbb{E}[X])^2.
$$

Substituting the values,

$$
=
\left[
\left(\frac{1}{36}\cdot 0.6\right)
+
\left(\frac{1}{4}\cdot 0.4\right)
\right]
-
(0.3)^2.
$$

$$
=
(0.0167 + 0.1) - 0.09
=
0.0267.
$$

# Final Decomposition

Therefore,

$$
0.21
=
0.1833
+
0.0267.
$$

That is,

$$
\text{Total Error}
=
\text{Remaining Error}
+
\text{Error Reduction}.
$$

# Summary

The Law of Total Variance shows that the optimal prediction error is always less than or equal to the global prediction error.

The quantity

$$
\mathrm{Var}(\mathbb{E}[X\mid Y])
$$

measures the predictive power of \(Y\). In this example, it quantifies exactly how much the bank improves its predictions by incorporating the risk class information.
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Joint Multivariate Normal Distribution

Let

$$
\mathbf{Z} =
\begin{bmatrix}
\mathbf{X} \\
\mathbf{Y}
\end{bmatrix}
\in \mathbb{R}^{n+m}.
$$

The random variable \(\mathbf{Z}\) is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

---

## Condition for a Joint Density

For \(\mathbf{Z}\) to possess a **joint density function** (i.e., be absolutely continuous with respect to the Lebesgue measure on \(\mathbb{R}^{n+m}\)), its covariance matrix must be **non-singular (positive definite)**.

---

## Joint Density Function

The joint density function of \(\mathbf{X}, \mathbf{Y}\) is

$$
f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})
=
\frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}}
\exp\left(
-\frac{1}{2}
(\mathbf{z}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{z}-\boldsymbol{\mu})
\right),
$$

where \(\mathbf{z} \in \mathbb{R}^{n+m}\).

---

## Parameters

### Mean Vector

$$
\boldsymbol{\mu}
=
\mathbb{E}[\mathbf{Z}]
=
\begin{bmatrix}
\boldsymbol{\mu}_X \\
\boldsymbol{\mu}_Y
\end{bmatrix}
\in \mathbb{R}^{n+m},
$$

where:
- \(\boldsymbol{\mu}_X \in \mathbb{R}^n\)
- \(\boldsymbol{\mu}_Y \in \mathbb{R}^m\)

---

### Covariance Matrix

$$
\boldsymbol{\Sigma}
=
\begin{bmatrix}
\boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\
\boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY}
\end{bmatrix}
$$

where:

- \(\boldsymbol{\Sigma}_{XX} = \mathrm{Var}(\mathbf{X}) \in \mathbb{R}^{n \times n}\)
- \(\boldsymbol{\Sigma}_{YY} = \mathrm{Var}(\mathbf{Y}) \in \mathbb{R}^{m \times m}\)
- \(\boldsymbol{\Sigma}_{XY} = \mathrm{Cov}(\mathbf{X}, \mathbf{Y}) \in \mathbb{R}^{n \times m}\)
- \(\boldsymbol{\Sigma}_{YX} = \boldsymbol{\Sigma}_{XY}^T\)

---

### Determinant

$$
|\boldsymbol{\Sigma}|
$$

is the normalization factor ensuring the density integrates to 1 over \(\mathbb{R}^{n+m}\).
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Marginalization of a Joint Gaussian via Block Matrix Methods

To derive the marginal density of \(\mathbf{X}\) (or \(\mathbf{Y}\)) from the joint density of \((\mathbf{X}, \mathbf{Y})\), we integrate the joint density over the remaining variable.

---

## Key Idea

The joint density has the form

$$
f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})
\propto
\exp\left(-\frac{1}{2}Q\right),
$$

where the quadratic form is

$$
Q =
\begin{bmatrix}
\mathbf{x}-\boldsymbol{\mu}_x \\
\mathbf{y}-\boldsymbol{\mu}_y
\end{bmatrix}^T
\boldsymbol{\Sigma}^{-1}
\begin{bmatrix}
\mathbf{x}-\boldsymbol{\mu}_x \\
\mathbf{y}-\boldsymbol{\mu}_y
\end{bmatrix}.
$$

---

## Block Covariance Structure

The covariance matrix is partitioned as

$$
\boldsymbol{\Sigma}
=
\begin{bmatrix}
\boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\
\boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY}
\end{bmatrix}.
$$

---

## Schur Complement Decomposition

Using block matrix inversion (Schur complement), we can express the inverse as

$$
\boldsymbol{\Sigma}^{-1}
=
\begin{bmatrix}
\mathbf{I} & -\boldsymbol{\Sigma}_{XX}^{-1}\boldsymbol{\Sigma}_{XY} \\
\mathbf{0} & \mathbf{I}
\end{bmatrix}^T
\begin{bmatrix}
\boldsymbol{\Sigma}_{XX}^{-1} & \mathbf{0} \\
\mathbf{0} & \mathbf{S}_{YY}^{-1}
\end{bmatrix}
\begin{bmatrix}
\mathbf{I} & -\boldsymbol{\Sigma}_{XX}^{-1}\boldsymbol{\Sigma}_{XY} \\
\mathbf{0} & \mathbf{I}
\end{bmatrix},
$$

where the Schur complement is

$$
\mathbf{S}_{YY}
=
\boldsymbol{\Sigma}_{YY}
-
\boldsymbol{\Sigma}_{YX}
\boldsymbol{\Sigma}_{XX}^{-1}
\boldsymbol{\Sigma}_{XY}.
$$

---

## Interpretation

- The Schur complement \(\mathbf{S}_{YY}\) represents the **conditional covariance structure** of \(\mathbf{Y}\) given \(\mathbf{X}\).
- This decomposition is the key step used in:
  - completing the square in the exponent
  - deriving marginal and conditional Gaussian distributions
  - proving closure properties of multivariate normal distributions

---
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Marginal Distribution of \(\mathbf{X}\) from a Joint Gaussian

Let

$$
\widetilde{\mathbf{x}} = \mathbf{x} - \boldsymbol{\mu}_x,
\qquad
\widetilde{\mathbf{y}} = \mathbf{y} - \boldsymbol{\mu}_y.
$$

We decompose the quadratic form \(Q\) as

$$
Q =
\widetilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{XX}^{-1} \widetilde{\mathbf{x}}
+
(\widetilde{\mathbf{y}} - \mathbf{A}\widetilde{\mathbf{x}})^T
\mathbf{S}_{YY}^{-1}
(\widetilde{\mathbf{y}} - \mathbf{A}\widetilde{\mathbf{x}}),
$$

where

$$
\mathbf{A} = \boldsymbol{\Sigma}_{YX}\boldsymbol{\Sigma}_{XX}^{-1}.
$$

---

## Factorization of the Joint Density

The joint density becomes

$$
f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})
=
C \cdot
\exp\left(-\frac{1}{2}\widetilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{XX}^{-1}\widetilde{\mathbf{x}}\right)
\cdot
\exp\left(-\frac{1}{2}(\widetilde{\mathbf{y}}-\mathbf{m})^T \mathbf{S}_{YY}^{-1}(\widetilde{\mathbf{y}}-\mathbf{m})\right),
$$

where

$$
\mathbf{m} = \mathbf{A}\widetilde{\mathbf{x}}.
$$

---

## Marginal Density of \(\mathbf{X}\)

By definition,

$$
f_{\mathbf{X}}(\mathbf{x})
=
\int_{\mathbb{R}^m} f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})\,d\mathbf{y}.
$$

Since the first exponential term does not depend on \(\mathbf{y}\), we factor it out:

$$
f_{\mathbf{X}}(\mathbf{x})
=
C \cdot
\exp\left(-\frac{1}{2}\widetilde{\mathbf{x}}^T \boldsymbol{\Sigma}_{XX}^{-1}\widetilde{\mathbf{x}}\right)
\int_{\mathbb{R}^m}
\exp\left(
-\frac{1}{2}(\widetilde{\mathbf{y}}-\mathbf{m})^T \mathbf{S}_{YY}^{-1}(\widetilde{\mathbf{y}}-\mathbf{m})
\right)
d\mathbf{y}.
$$

---

## Gaussian Integral Result

The integral is the normalization constant of a multivariate Gaussian:

$$
\int_{\mathbb{R}^m} (\cdots)\, d\mathbf{y}
=
\sqrt{(2\pi)^m |\mathbf{S}_{YY}|}.
$$

---

## Determinant Identity

For block matrices,

$$
|\boldsymbol{\Sigma}|
=
|\boldsymbol{\Sigma}_{XX}|\cdot|\mathbf{S}_{YY}|.
$$

---

## Final Simplification

Substituting and simplifying constants yields

$$
f_{\mathbf{X}}(\mathbf{x})
=
\frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{XX}|}}
\exp\left(
-\frac{1}{2}
(\mathbf{x}-\boldsymbol{\mu}_x)^T
\boldsymbol{\Sigma}_{XX}^{-1}
(\mathbf{x}-\boldsymbol{\mu}_x)
\right).
$$

---

## Final Result

$$
\mathbf{X} \sim \mathcal{N}(\boldsymbol{\mu}_x,\boldsymbol{\Sigma}_{XX}).
$$

This shows that any marginal of a multivariate normal distribution is again Gaussian, with mean and covariance given by the corresponding sub-blocks of the full covariance matrix.
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Conditional Variance and the Law of Total Variance (Gaussian Case)

Since

$$
(\mathbf{X} \mid \mathbf{Y}=\mathbf{y})
\sim
\mathcal{N}\Big(
\boldsymbol{\mu}_x
+
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y}-\boldsymbol{\mu}_y),
\;
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}
\Big),
$$

we immediately obtain the conditional variance:

$$
\mathrm{Var}(\mathbf{X}\mid \mathbf{Y}=y)
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

---

## Independence of \(y\)

Since this expression does not depend on \(y\), we have

$$
\mathrm{Var}(\mathbf{X}\mid \mathbf{Y})
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

and therefore

$$
\mathbb{E}[\mathrm{Var}(\mathbf{X}\mid \mathbf{Y})]
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

---

## Estimation Error (MMSE)

Thus the optimal estimation error is

$$
\mathbb{E}\left[(\mathbf{X}-\mathbb{E}[\mathbf{X}\mid \mathbf{Y}])^2\right]
=
\mathbb{E}[\mathrm{Var}(\mathbf{X}\mid \mathbf{Y})]
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

---

## Law of Total Variance

For any random variables with finite variance,

$$
\mathrm{Var}(\mathbf{X})
=
\mathbb{E}[\mathrm{Var}(\mathbf{X}\mid \mathbf{Y})]
+
\mathrm{Var}(\mathbb{E}[\mathbf{X}\mid \mathbf{Y}]).
$$

---

## Decomposition in the Gaussian Case

Substituting the Gaussian conditional structure gives:

### Explained Variance

$$
\mathrm{Var}(\mathbb{E}[\mathbf{X}\mid \mathbf{Y}])
=
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

### Unexplained Variance

$$
\mathrm{Var}(\mathbf{X}\mid \mathbf{Y})
=
\boldsymbol{\Sigma}_{XX}
-
\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}.
$$

---

## Final Interpretation

- **Explained Variance:**
  $$
  \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}
  $$
  This is the part of the variability in \(\mathbf{X}\) that can be predicted from \(\mathbf{Y}\).

- **Unexplained Variance (MMSE):**
  $$
  \boldsymbol{\Sigma}_{XX}
  -
  \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}
  $$
  This is the irreducible estimation error after observing \(\mathbf{Y}\).

---
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Differential Entropy of a Multivariate Gaussian

Consider the multivariate normal random variable

$$
\mathbf{Z} =
\begin{bmatrix}
\mathbf{X} \\
\mathbf{Y}
\end{bmatrix}
\in \mathbb{R}^{n+m},
$$

with mean \(\boldsymbol{\mu} \in \mathbb{R}^{n+m}\) and covariance matrix
\(\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m)\times(n+m)}\).

The joint density is

$$
f_{\mathbf{X},\mathbf{Y}}(\mathbf{z})
=
\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol{\Sigma}|}}
\exp\left(
-\frac{1}{2}
(\mathbf{z}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{z}-\boldsymbol{\mu})
\right).
$$

---

# Differential Entropy

The differential entropy is defined as

$$
h(\mathbf{Z}) = -\mathbb{E}[\ln f_{\mathbf{X},\mathbf{Y}}(\mathbf{Z})].
$$

---

## Step A: Log-density

$$
\ln f_{\mathbf{X},\mathbf{Y}}(\mathbf{z})
=
-\frac{1}{2}\ln\left((2\pi)^{n+m}|\boldsymbol{\Sigma}|\right)
-\frac{1}{2}
(\mathbf{z}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{z}-\boldsymbol{\mu}).
$$

---

## Step B: Take expectation

$$
h(\mathbf{Z})
=
\frac{1}{2}\ln\left((2\pi)^{n+m}|\boldsymbol{\Sigma}|\right)
+
\frac{1}{2}
\mathbb{E}\left[
(\mathbf{Z}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{Z}-\boldsymbol{\mu})
\right].
$$

---

## Step C: Quadratic form expectation

For a multivariate Gaussian,

$$
\mathbb{E}\left[
(\mathbf{Z}-\boldsymbol{\mu})^T
\boldsymbol{\Sigma}^{-1}
(\mathbf{Z}-\boldsymbol{\mu})
\right]
=
\mathrm{Tr}(\boldsymbol{\Sigma}^{-1}\boldsymbol{\Sigma})
=
\mathrm{Tr}(\mathbf{I}_{n+m})
=
n+m.
$$

---

## Step D: Final expression

Substituting,

$$
h(\mathbf{Z})
=
\frac{1}{2}\ln\left((2\pi)^{n+m}|\boldsymbol{\Sigma}|\right)
+
\frac{n+m}{2}.
$$

Rewriting \(\frac{n+m}{2} = \frac{1}{2}\ln(e^{n+m})\),

---

## Final Result

$$
h(\mathbf{Z})
=
\frac{1}{2}
\ln\left((2\pi e)^{n+m}|\boldsymbol{\Sigma}|\right).
$$

---

## Key Insight

The entropy of a multivariate Gaussian depends only on:
- dimension \(n+m\)
- determinant of covariance \(|\boldsymbol{\Sigma}|\)

It is **maximized among all distributions with the same covariance**, making the Gaussian the maximum entropy distribution.
"""))
```

```python
# Google Colab / Jupyter Notebook Display Code
from IPython.display import display, Markdown

display(Markdown(r"""
# Joint Gaussian Model: Signal + Noise Interpretation

---

## 1. Joint Distribution of \((X, Y)\)

Let
\[
Y = X + Q,
\]
where \(X \sim \mathcal{N}(0,1)\), \(Q \sim \mathcal{N}(0,4)\), and \(X \perp Q\).

### Mean
\[
\mathbb{E}[Y] = \mathbb{E}[X] + \mathbb{E}[Q] = 0.
\]

### Variance
\[
\mathrm{Var}(Y) = \mathrm{Var}(X) + \mathrm{Var}(Q) = 1 + 4 = 5.
\]

### Covariance
\[
\mathrm{Cov}(X,Y)
=
\mathbb{E}[X(X+Q)]
=
\mathbb{E}[X^2] + \mathbb{E}[XQ]
= 1 + 0 = 1.
\]

### Joint distribution
\[
\begin{pmatrix}
X \\
Y
\end{pmatrix}
\sim
\mathcal{N}
\left(
\begin{pmatrix}
0 \\
0
\end{pmatrix},
\begin{pmatrix}
1 & 1 \\
1 & 5
\end{pmatrix}
\right).
\]

---

## 2. Information Structure

\(X\) is not a function of \(Y\). Many \((X,Q)\) pairs produce the same \(Y\), so the system is not invertible.

Interpretation:
\(Y\) mixes signal (\(X\)) and noise (\(Q\)), so it cannot perfectly reveal \(X\).

---

## 3. Conditional Expectation (MMSE Estimator)

For jointly Gaussian variables:

\[
\mathbb{E}[X \mid Y=y]
=
\mu_X + \frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}(y-\mu_Y).
\]

Substitute values:

\[
\mathbb{E}[X \mid Y=y]
=
\frac{1}{5}y.
\]

So,
\[
\mathbb{E}[X \mid Y] = 0.2Y.
\]

---

## 4. Residual (Estimation Error)

Let
\[
Z = X - 0.2Y.
\]

### Mean
\[
\mathbb{E}[Z] = 0.
\]

### Orthogonality
\[
\mathrm{Cov}(Z,Y)
=
\mathbb{E}[XY] - 0.2\mathbb{E}[Y^2]
=
1 - 0.2 \cdot 5
=
0.
\]

So the error is uncorrelated with the observation.

---

## 5. Differential Entropy

For a Gaussian:
\[
h(\mathcal{N}(0,\sigma^2)) = \frac{1}{2}\ln(2\pi e \sigma^2).
\]

### Marginal entropy
\[
h(X) = \frac{1}{2}\ln(2\pi e).
\]

### Conditional variance
\[
\mathrm{Var}(X|Y)
=
1 - \frac{1}{5}
=
0.8.
\]

### Conditional entropy
\[
h(X|Y) = \frac{1}{2}\ln(2\pi e \cdot 0.8).
\]

---

## 6. Mutual Information

\[
I(X;Y) = h(X) - h(X|Y)
= \frac{1}{2}\ln\left(\frac{1}{0.8}\right)
= \frac{1}{2}\ln(1.25).
\]

\[
I(X;Y) \approx 0.1116 \text{ nats}.
\]

Interpretation:
- Only a small fraction of uncertainty in \(X\) is explained by \(Y\).
- Noise dominates signal moderately.

---

## 7. Estimation Insight

### Naive estimator
\[
\hat{X} = Y \quad \Rightarrow \quad \mathbb{E}[(X-Y)^2] = 4.
\]

### Optimal estimator
\[
\hat{X} = 0.2Y \quad \Rightarrow \quad \mathrm{MSE} = 0.8.
\]

---

## Key Insight (Shrinkage)

The factor \(0.2\) shows **regression to the mean**:
- Do not fully trust noisy observations
- Optimal estimator shrinks toward expectation

This is the essence of Bayesian estimation and Kalman filtering.
"""))
```

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import multivariate_normal

# --- 1. Model Parameters ---
mu_x, mu_y = 0, 0
var_x, var_q = 1, 4
var_y = var_x + var_q
cov_xy = var_x

k = cov_xy / var_y  # 0.2

# --- 2. Generate Samples ---
np.random.seed(42)
n_samples = 1500

x_samples = np.random.normal(0, np.sqrt(var_x), n_samples)
q_samples = np.random.normal(0, np.sqrt(var_q), n_samples)
y_samples = x_samples + q_samples

x_hat = k * y_samples
residual = x_samples - x_hat

# --- 3. Grid for Joint Density ---
x_grid = np.linspace(-4, 4, 200)
y_grid = np.linspace(-6, 6, 200)

X_m, Y_m = np.meshgrid(x_grid, y_grid)
pos = np.dstack((X_m, Y_m))

cov_matrix = np.array([[var_x, cov_xy],
                       [cov_xy, var_y]])

Z = multivariate_normal([0, 0], cov_matrix).pdf(pos)

# --- 4. Plot ---
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=(
        "Joint Distribution (X, Y) with Regression Line",
        "Residual vs Observation (Orthogonality)"
    )
)

# --- LEFT: Joint distribution ---
fig.add_trace(
    go.Contour(
        x=x_grid,
        y=y_grid,
        z=Z,
        colorscale='Blues',
        opacity=0.7,
        showscale=False
    ),
    row=1, col=1
)

# Scatter samples
fig.add_trace(
    go.Scatter(
        x=x_samples,
        y=y_samples,
        mode='markers',
        marker=dict(size=3, opacity=0.25),
        name="Samples"
    ),
    row=1, col=1
)

# Optimal estimator line: x = 0.2y
y_line = np.linspace(-6, 6, 100)

fig.add_trace(
    go.Scatter(
        x=k * y_line,
        y=y_line,
        line=dict(color='red', width=4),
        name="E[X|Y]=0.2Y"
    ),
    row=1, col=1
)

# Naive estimator x = y
fig.add_trace(
    go.Scatter(
        x=y_line,
        y=y_line,
        line=dict(color='black', dash='dash', width=2),
        name="Naive: X=Y"
    ),
    row=1, col=1
)

# --- RIGHT: Residual ---
fig.add_trace(
    go.Scatter(
        x=y_samples,
        y=residual,
        mode='markers',
        marker=dict(size=4, opacity=0.4, color='purple'),
        name="Residual"
    ),
    row=1, col=2
)

fig.add_trace(
    go.Scatter(
        x=y_line,
        y=np.zeros_like(y_line),
        line=dict(color='black'),
        name="Zero line"
    ),
    row=1, col=2
)

# --- Layout ---
fig.update_layout(
    height=600,
    width=1100,
    title="Gaussian Signal Extraction: MMSE Estimation and Orthogonality",
    template="plotly_white"
)

fig.update_xaxes(title_text="X", row=1, col=1)
fig.update_yaxes(title_text="Y", row=1, col=1)

fig.update_xaxes(title_text="Y", row=1, col=2)
fig.update_yaxes(title_text="Residual (X - 0.2Y)", row=1, col=2)

fig.show()
```
