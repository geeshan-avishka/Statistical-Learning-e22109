Q1 Statistical Mechanics of the Quantum Oscillator

```python
from IPython.display import display, Markdown, Math

display(Markdown("# Quantum Harmonic Oscillator – Canonical Ensemble"))

display(Markdown("## 1. Partition Function"))

display(Math(r"E_j=\hbar\omega\left(j+\frac12\right),\qquad j=0,1,2,\ldots"))

display(Math(r"Z(T)=\sum_{j=0}^{\infty}e^{-E_j/(k_BT)}"))

display(Math(r"Z=e^{-\beta\hbar\omega/2}\sum_{j=0}^{\infty}\left(e^{-\beta\hbar\omega}\right)^j"))

display(Math(r"\beta=\frac{1}{k_BT}"))

display(Math(r"\sum_{j=0}^{\infty}r^j=\frac{1}{1-r}"))

display(Math(r"\boxed{Z(T)=\frac{e^{-\frac{\hbar\omega}{2k_BT}}}{1-e^{-\frac{\hbar\omega}{k_BT}}}}"))

display(Markdown("---"))

display(Markdown("## 2. Helmholtz Free Energy"))

display(Math(r"\ln Z=-\frac{\beta\hbar\omega}{2}-\ln\left(1-e^{-\beta\hbar\omega}\right)"))

display(Math(r"\boxed{F=\frac{\hbar\omega}{2}+k_BT\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)}"))

display(Markdown("### Internal Energy"))

display(Math(r"U=-\frac{\partial\ln Z}{\partial\beta}"))

display(Math(r"\boxed{U=\frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_BT}}-1}}"))

display(Markdown("The first term is the zero-point energy and the second term is the thermal excitation energy."))

display(Markdown("---"))

display(Markdown("## 3. Entropy"))

display(Math(r"S=-\left(\frac{\partial F}{\partial T}\right)"))

display(Math(r"\boxed{S=k_B\left[\frac{\frac{\hbar\omega}{k_BT}}{e^{\frac{\hbar\omega}{k_BT}}-1}-\ln\left(1-e^{-\frac{\hbar\omega}{k_BT}}\right)\right]}"))

display(Markdown("### Verification"))

display(Math(r"x=\frac{\hbar\omega}{k_BT}"))

display(Math(r"F+TS=\frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^x-1}=U"))

display(Math(r"\boxed{U=F+TS}"))

display(Markdown("---"))

display(Markdown("## 4. Heat Capacity"))

display(Math(r"C_V=\left(\frac{\partial U}{\partial T}\right)"))

display(Math(r"\boxed{C_V=k_B\left(\frac{\hbar\omega}{k_BT}\right)^2\frac{e^{\frac{\hbar\omega}{k_BT}}}{\left(e^{\frac{\hbar\omega}{k_BT}}-1\right)^2}}"))

display(Markdown("### Hyperbolic Form"))

display(Math(r"\sinh(y)=\frac{e^y-e^{-y}}{2}"))

display(Math(r"y=\frac{\hbar\omega}{2k_BT}"))

display(Math(r"\frac{e^x}{(e^x-1)^2}=\frac{1}{4\sinh^2(y)}"))

display(Math(r"\boxed{C_V=k_B\left(\frac{\hbar\omega}{2k_BT}\right)^2\frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_BT}\right)}}"))

display(Markdown("---"))

display(Markdown("## 5. High-Temperature Limit"))

display(Math(r"x=\frac{\hbar\omega}{k_BT}\ll1"))

display(Math(r"\sinh\left(\frac{x}{2}\right)\approx\frac{x}{2}"))

display(Math(r"C_V\approx k_B\left(\frac{x}{2}\right)^2\frac{1}{(x/2)^2}=k_B"))

display(Math(r"\boxed{\lim_{T\rightarrow\infty}C_V=k_B}"))

display(Markdown(
"""
At high temperatures, the thermal energy is much larger than the energy spacing between quantum levels. The oscillator behaves classically, and the heat capacity approaches **\(k_B\)**. This exactly agrees with the classical equipartition theorem, which predicts an average energy of **\(k_BT\)** for a one-dimensional harmonic oscillator and therefore **\(C_V=k_B\)**.
"""
))
```

Q2. The Maximum Entropy Distribution under Mean and Variance Constraints

```python
from IPython.display import display, Markdown, Math

display(Markdown("# Maximum Entropy Principle with Mean and Variance Constraints"))

display(Markdown("## Given"))

display(Math(r"\mathbb{E}[X]=\mu"))

display(Math(r"\mathbb{E}\left[(X-\mu)^2\right]=\sigma^2"))

display(Markdown(
"""
We seek the probability density function \(p(x)\) that maximizes the differential entropy while satisfying the above constraints.
"""
))

display(Markdown("---"))

display(Markdown("## 1. Differential Entropy"))

display(Math(r"H[p]=-\int_{-\infty}^{\infty}p(x)\ln p(x)\,dx"))

display(Markdown(
"""
The constraints are:
"""
))

display(Math(r"\int_{-\infty}^{\infty}p(x)\,dx=1"))

display(Math(r"\int_{-\infty}^{\infty}xp(x)\,dx=\mu"))

display(Math(r"\int_{-\infty}^{\infty}(x-\mu)^2p(x)\,dx=\sigma^2"))

display(Markdown("---"))

display(Markdown("## 2. Construct the Lagrangian"))

display(Math(
r"\mathcal{L}=-\int p(x)\ln p(x)\,dx-\lambda_0\left(\int p(x)\,dx-1\right)-\lambda_1\left(\int xp(x)\,dx-\mu\right)-\lambda_2\left(\int (x-\mu)^2p(x)\,dx-\sigma^2\right)"
))

display(Markdown("---"))

display(Markdown("## 3. Maximize the Entropy"))

display(Markdown(
"""
Taking the functional derivative with respect to \(p(x)\) and setting it equal to zero,
"""
))

display(Math(r"\frac{\delta\mathcal{L}}{\delta p(x)}=0"))

display(Math(r"-(\ln p(x)+1)-\lambda_0-\lambda_1x-\lambda_2(x-\mu)^2=0"))

display(Math(r"\ln p(x)=-1-\lambda_0-\lambda_1x-\lambda_2(x-\mu)^2"))

display(Math(r"p(x)=Ae^{-\lambda_1x-\lambda_2(x-\mu)^2}"))

display(Markdown(
"""
Since the distribution is centered at the known mean, the linear term vanishes after applying the constraints, giving
"""
))

display(Math(r"p(x)=Ae^{-\lambda_2(x-\mu)^2}"))

display(Markdown("---"))

display(Markdown("## 4. Determine the Constants"))

display(Markdown(
"""
Using the normalization condition,
"""
))

display(Math(r"\int_{-\infty}^{\infty}p(x)\,dx=1"))

display(Math(r"A=\frac{1}{\sqrt{2\pi\sigma^2}}"))

display(Markdown(
"""
Using the variance constraint,
"""
))

display(Math(r"\lambda_2=\frac{1}{2\sigma^2}"))

display(Markdown("---"))

display(Markdown("## 5. Maximum Entropy Distribution"))

display(Math(
r"\boxed{p(x)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)}"
))

display(Markdown("---"))

display(Markdown("## Conclusion"))

display(Markdown(
"""
The probability density function that maximizes the differential entropy among all continuous probability distributions having a fixed mean **μ** and variance **σ²** is the **Gaussian (Normal) distribution**.

Therefore,

"""
))

display(Math(
r"\boxed{X\sim\mathcal{N}(\mu,\sigma^2)}"
))

display(Markdown(
"""
Hence, the maximum entropy law is

\[
X\sim\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right),
\]

which is exactly the required result.
"""
))
```

Q3. Multiclass Logistic Regression: Automatic Classification of Emails

```python
from IPython.display import display, Markdown, Math

display(Markdown("# Maximum Entropy Classification Solution"))

display(Markdown("## Problem Statement"))

display(Markdown(
"""
The target variable is

- **Y = 0** : Finance
- **Y = 1** : Technical

The AI Feature Extractor outputs the binary feature vector

\[
\mathbf{x}=(x_1,x_2,x_3)^T
\]

where

- \(x_1\): Financial context detected
- \(x_2\): Account access context detected
- \(x_3\): Bug or malfunction context detected
"""
))

display(Markdown("---"))

display(Markdown("## 1. Feature States"))

display(Math(r"\mathbf{x}_A=(1,0,0)^T,\qquad P(\mathbf{x}_A)=0.40"))

display(Math(r"\mathbf{x}_B=(0,1,0)^T,\qquad P(\mathbf{x}_B)=0.25"))

display(Math(r"\mathbf{x}_C=(0,0,1)^T,\qquad P(\mathbf{x}_C)=0.05"))

display(Math(r"\mathbf{x}_D=(0,1,1)^T,\qquad P(\mathbf{x}_D)=0.15"))

display(Math(r"\mathbf{x}_E=(1,1,0)^T,\qquad P(\mathbf{x}_E)=0.05"))

display(Markdown("The total observed probability is"))

display(Math(r"0.40+0.25+0.05+0.15+0.05=0.90"))

display(Markdown("Hence, 10% probability corresponds to other unlisted feature combinations."))

display(Markdown("---"))

display(Markdown("## 2. Maximum Entropy Model"))

display(Markdown(
"""
The conditional probability distribution that maximizes entropy subject to the feature expectation constraints belongs to the exponential family.
"""
))

display(Math(
r"P(Y=y|\mathbf{x})=\frac{\exp\left(\sum_i\lambda_i f_i(\mathbf{x},y)\right)}{Z(\mathbf{x})}"
))

display(Math(
r"Z(\mathbf{x})=\sum_y\exp\left(\sum_i\lambda_i f_i(\mathbf{x},y)\right)"
))

display(Markdown("For binary classification, the model becomes"))

display(Math(
r"P(Y=1|\mathbf{x})=\frac{\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}{1+\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}"
))

display(Math(
r"P(Y=0|\mathbf{x})=\frac{1}{1+\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}"
))

display(Markdown("---"))

display(Markdown("## 3. Given Expectation Constraints"))

display(Math(r"\mathbb{E}[x_1\mathbb{I}(Y=1)]=0.05"))

display(Math(r"\mathbb{E}[x_2\mathbb{I}(Y=1)]=0.15"))

display(Math(r"\mathbb{E}[x_3\mathbb{I}(Y=1)]=0.16"))

display(Markdown(
"""
These expectations determine the unknown parameters

\[
\lambda_1,\lambda_2,\lambda_3
\]

through maximum likelihood (or Generalized Iterative Scaling / Improved Iterative Scaling).
"""
))

display(Markdown("---"))

display(Markdown("## 4. Interpretation of the Constraints"))

display(Markdown(
"""
### Constraint 1

\[
\mathbb{E}[x_1\mathbb{I}(Y=1)]=0.05
\]

Financial-related emails are rarely routed to Technical support.

Therefore,

\[
\boxed{\lambda_1<0}
\]

indicating a negative contribution toward the Technical class.
"""
))

display(Markdown(
"""
### Constraint 2

\[
\mathbb{E}[x_2\mathbb{I}(Y=1)]=0.15
\]

Account-access issues are moderately associated with Technical support.

Therefore,

\[
\boxed{\lambda_2>0}
\]

with moderate magnitude.
"""
))

display(Markdown(
"""
### Constraint 3

\[
\mathbb{E}[x_3\mathbb{I}(Y=1)]=0.16
\]

Bug or malfunction features are strongly associated with Technical support.

Therefore,

\[
\boxed{\lambda_3>0}
\]

and typically

\[
|\lambda_3|>|\lambda_2|.
\]
"""
))

display(Markdown("---"))

display(Markdown("## 5. Resulting Maximum Entropy Classifier"))

display(Math(
r"\boxed{P(Y=1|\mathbf{x})=\frac{\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}{1+\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}}"
))

display(Math(
r"\boxed{P(Y=0|\mathbf{x})=\frac{1}{1+\exp(\lambda_1x_1+\lambda_2x_2+\lambda_3x_3)}}"
))

display(Markdown(
"""
The learned parameters satisfy

\[
\boxed{\lambda_1<0,\qquad \lambda_2>0,\qquad \lambda_3>0,\qquad |\lambda_3|>|\lambda_2|}
\]

which is consistent with the observed feature expectations.
"""
))

display(Markdown("---"))

display(Markdown("## Conclusion"))

display(Markdown(
"""
Using the Maximum Entropy Principle, the conditional distribution is modeled by an exponential (logistic) classifier. The model satisfies all empirical expectation constraints while remaining the least biased distribution consistent with the available information.

The resulting classifier correctly reflects the historical routing behavior:

- Financial keywords strongly favor **Finance**.
- Account-access keywords moderately favor **Technical**.
- Bug or malfunction keywords strongly favor **Technical**.

Thus, the Maximum Entropy model provides the probability of routing each incoming email to the Technical or Finance department while preserving maximum uncertainty subject to the observed constraints.
"""
))
```

Q4. Asset Pricing and Portfolio Optimization.

```python
from IPython.display import display, Markdown, Math

display(Markdown(r"# Maximum Entropy Solution for Risk-Neutral Asset Price Distribution"))

display(Markdown(r"""
## Problem Statement

Let the future asset price after one month be denoted by

\[
S_T>0.
\]

The objective is to determine the probability density function \(p(S_T)\) that maximizes entropy while satisfying the observed market constraints.
"""))

display(Markdown("---"))

display(Markdown("## 1. Given Market Constraints"))

display(Math(r"\mathbb{E}[S_T]=102"))
display(Math(r"\mathbb{E}[S_T^2]=10600"))
display(Math(r"\mathbb{E}[\ln(S_T)]=4.60"))

display(Markdown(r"""
The density must also satisfy the normalization condition.
"""))

display(Math(r"\int_{0}^{\infty}p(S_T)\,dS_T=1"))

display(Markdown("---"))

display(Markdown("## 2. Differential Entropy"))

display(Math(r"H[p]=-\int_{0}^{\infty}p(S_T)\ln p(S_T)\,dS_T"))

display(Markdown(r"""
The Maximum Entropy Principle chooses the probability density that maximizes this entropy while satisfying all known market information.
"""))

display(Markdown("---"))

display(Markdown("## 3. Construct the Lagrangian"))

display(Math(r"\mathcal{L}=-\int p(S_T)\ln p(S_T)\,dS_T"))

display(Math(r"-\lambda_0\left(\int p(S_T)\,dS_T-1\right)"))

display(Math(r"-\lambda_1\left(\int S_Tp(S_T)\,dS_T-102\right)"))

display(Math(r"-\lambda_2\left(\int S_T^2p(S_T)\,dS_T-10600\right)"))

display(Math(r"-\lambda_3\left(\int \ln(S_T)p(S_T)\,dS_T-4.60\right)"))

display(Markdown("---"))

display(Markdown("## 4. Maximize the Entropy"))

display(Markdown(r"""
Taking the functional derivative with respect to \(p(S_T)\) and setting it equal to zero,
"""))

display(Math(r"\frac{\delta\mathcal{L}}{\delta p(S_T)}=0"))

display(Math(r"-(\ln p(S_T)+1)-\lambda_0-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)=0"))

display(Math(r"\ln p(S_T)=-1-\lambda_0-\lambda_1S_T-\lambda_2S_T^2-\lambda_3\ln(S_T)"))

display(Markdown(r"Exponentiating both sides gives"))

display(Math(r"p(S_T)=A\,S_T^{-\lambda_3}\exp\left(-\lambda_1S_T-\lambda_2S_T^2\right)"))

display(Markdown(r"""
where

\[
A=e^{-1-\lambda_0}
\]

is the normalization constant.
"""))

display(Markdown("---"))

display(Markdown("## 5. Normalization Constant"))

display(Math(r"\int_{0}^{\infty}A\,S_T^{-\lambda_3}\exp\left(-\lambda_1S_T-\lambda_2S_T^2\right)dS_T=1"))

display(Math(r"A=\left[\int_{0}^{\infty}S_T^{-\lambda_3}\exp\left(-\lambda_1S_T-\lambda_2S_T^2\right)dS_T\right]^{-1}"))

display(Markdown(r"""
The Lagrange multipliers

\[
\lambda_1,\lambda_2,\lambda_3
\]

are determined numerically so that all three market constraints are exactly satisfied.
"""))

display(Markdown("---"))

display(Markdown("## 6. Final Maximum Entropy Distribution"))

display(Math(r"\boxed{p(S_T)=A\,S_T^{-\lambda_3}\exp\left(-\lambda_1S_T-\lambda_2S_T^2\right),\qquad S_T>0}"))

display(Markdown(r"""
This is the least-biased continuous probability density consistent with all available market information.
"""))

display(Markdown("---"))

display(Markdown("## 7. Interpretation of the Constraints"))

display(Markdown(r"""
### Expected Price

\[
\mathbb{E}[S_T]=102
\]

ensures the distribution is centered on the market-implied forward price.
"""))

display(Markdown(r"""
### Expected Squared Price

\[
\mathbb{E}[S_T^2]=10600
\]

controls the market-implied volatility and dispersion.
"""))

display(Markdown(r"""
### Logarithmic Constraint

\[
\mathbb{E}[\ln(S_T)]=4.60
\]

introduces asymmetry into the distribution and captures the downside crash risk implied by option prices.
"""))

display(Markdown("---"))

display(Markdown(r"""
## Conclusion

Using the Maximum Entropy Principle, the probability density of the future stock price is obtained without assuming a Normal distribution. Instead, the density is completely determined by the observed market constraints on the expected price, expected squared price, and expected logarithmic price.

The resulting distribution

\[
p(S_T)=A\,S_T^{-\lambda_3}\exp\left(-\lambda_1S_T-\lambda_2S_T^2\right)
\]

is the unique maximum entropy solution. It incorporates forward pricing, volatility, and downside market sentiment while remaining the least-biased probability distribution consistent with the available information.
"""))
```
