```python


from IPython.display import display, Markdown
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# =========================
# DISPLAY THEORY SECTION
# =========================

display(Markdown(r"""
# Beta Distribution Model for Defective Proportion

## 1. Probability Space

A suitable probability space is

$$
\Omega=[0,1], \qquad \mathfrak F=\mathfrak B([0,1]),
$$

where $\mathfrak B([0,1])$ is the Borel $\sigma$-algebra on $[0,1]$.

Define the probability measure $P$ by

$$
P(A)=\int_A f(x)\,dx, \qquad A\in \mathfrak F,
$$

where

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\\\
0, & \text{otherwise}.
\end{cases}
$$

---

## 2. Random Variable

Define the random variable $X:\Omega\to\mathbb R$ by

$$
X(\omega)=\omega, \qquad \omega\in[0,1].
$$

Thus $X$ records the defective proportion in the batch.

---

## 3. Probability Density Function

The probability density function of $X$ is

$$
f_X(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\\\
0, & \text{otherwise}.
\end{cases}
$$

---

## 4. Verification of Valid Density Function

To verify that $f$ is a valid density, note first that

$$
f(x)\ge 0
$$

for all $x$.

Also,

$$
\int_{-\infty}^{\infty} f(x)\,dx
=
\int_0^1 \dfrac{1}{B(2,18)}x(1-x)^{17}\,dx
$$

$$
=
\dfrac{1}{B(2,18)}B(2,18)
$$

$$
=1.
$$

So $f$ is a valid probability density function.

---

## 5. Expectation

Since

$$
X\sim \mathrm{Beta}(2,18),
$$

we have

$$
E[X]=\frac{\alpha}{\alpha+\beta}
$$

$$
=\frac{2}{20}
$$

$$
=0.10.
$$

---

## 6. Variance

The variance of a Beta$(\alpha,\beta)$ random variable is

$$
\mathrm{Var}(X)=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}.
$$

Hence

$$
\mathrm{Var}(X)=
\frac{2\cdot18}{20^2\cdot21}
$$

$$
=
\frac{36}{8400}
$$

$$
=
\frac{3}{700}
$$

$$
\approx 0.00429.
$$

---

## 7. Probability that $X \le 0.10$

$$
P(X\le 0.10)
=
\int_0^{0.10}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Equivalently,

$$
P(X\le 0.10)=I_{0.10}(2,18),
$$

where $I_x(\alpha,\beta)$ denotes the regularized incomplete Beta function.

---

## 8. Probability that $0.05 \le X \le 0.15$

$$
P(0.05\le X\le 0.15)
=
\int_{0.05}^{0.15}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx.
$$

Equivalently,

$$
P(0.05\le X\le 0.15)
=
I_{0.15}(2,18)-I_{0.05}(2,18).
$$

---

## 9. Why the Beta Distribution is Appropriate

A Beta distribution is reasonable here because the defective proportion is a continuous quantity constrained to lie between $0$ and $1$, and the Beta family is specifically designed to model proportions and rates.

---

## 10. Interpretation

The expectation

$$
E[X]=0.10
$$

means that the long-run average defective proportion is $10\%$.

The variance

$$
\mathrm{Var}(X)\approx 0.00429
$$

measures the variability in the defective proportion from batch to batch, indicating how stable or unstable the manufacturing quality is over time.
"""))

# =========================
# GRAPH SECTION
# =========================

# Parameters
alpha = 2
beta_param = 18

# x values
x = np.linspace(0, 1, 1000)

# Beta PDF
y = beta.pdf(x, alpha, beta_param)

# Graph 1: PDF Curve
plt.figure(figsize=(10,5))
plt.plot(x, y, linewidth=2)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Beta(2,18) Probability Density Function')
plt.grid(True)
plt.show()

# Graph 2: Shaded Probability P(X <= 0.10)
plt.figure(figsize=(10,5))
plt.plot(x, y, linewidth=2)

x_fill = np.linspace(0, 0.10, 500)
y_fill = beta.pdf(x_fill, alpha, beta_param)

plt.fill_between(x_fill, y_fill, alpha=0.5)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Probability Region: P(X ≤ 0.10)')
plt.grid(True)
plt.show()

# Graph 3: Shaded Probability 0.05 <= X <= 0.15
plt.figure(figsize=(10,5))
plt.plot(x, y, linewidth=2)

x_fill2 = np.linspace(0.05, 0.15, 500)
y_fill2 = beta.pdf(x_fill2, alpha, beta_param)

plt.fill_between(x_fill2, y_fill2, alpha=0.5)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Probability Region: 0.05 ≤ X ≤ 0.15')
plt.grid(True)
plt.show()
```

```python



from IPython.display import display, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang

# ---------------------------------------------------
# Display Mathematical Explanation
# ---------------------------------------------------

display(Markdown(r"""
# Triangular Distribution Model for Installation Time

## 1. Support of the Random Variable

Since \(X\) represents the installation time in hours, and the best-case and worst-case estimates are \(4\) and \(10\), the support of \(X\) is

$$
4 \leq X \leq 10.
$$

---

## 2. Probability Density Function

Since

$$
X \sim \mathrm{Triangular}(4,10,6),
$$

its density is

$$
f_X(x)=
\begin{cases}
\dfrac{2(x-4)}{(10-4)(6-4)}, & 4 \leq x < 6,\\[8pt]
\dfrac{2}{10-4}, & x=6,\\[8pt]
\dfrac{2(10-x)}{(10-4)(10-6)}, & 6 < x \leq 10, \\[8pt]
0, & \text{otherwise}.
\end{cases}
$$

This simplifies to

$$
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4 \leq x < 6,\\[8pt]
\dfrac{1}{3}, & x=6,\\[8pt]
\dfrac{10-x}{12}, & 6 < x \leq 10,\\[8pt]
0, & \text{otherwise}.
\end{cases}
$$

---

## 3. Verification of the Density Function

To verify that \(f_X\) is a valid density:

$$
\int_{4}^{6} \frac{x-4}{6}\,dx
+
\int_{6}^{10} \frac{10-x}{12}\,dx.
$$

For the first integral,

$$
\int_{4}^{6} \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_{4}^{6} (x-4)\,dx
=
\frac{1}{6}\cdot 2
=
\frac{1}{3}.
$$

For the second integral,

$$
\int_{6}^{10} \frac{10-x}{12}\,dx
=
\frac{1}{12}\int_{6}^{10} (10-x)\,dx
=
\frac{1}{12}\cdot 8
=
\frac{2}{3}.
$$

Hence,

$$
\int_{-\infty}^{\infty} f_X(x)\,dx
=
\frac{1}{3}+\frac{2}{3}
=
1.
$$

Therefore, \(f_X\) is a valid probability density function.

---

## 4. Expectation

The mean of a triangular distribution is

$$
E[X]=\frac{a+b+c}{3}.
$$

Thus,

$$
E[X]
=
\frac{4+10+6}{3}
=
\frac{20}{3}
\approx 6.67.
$$

---

## 5. Variance

The variance formula is

$$
\mathrm{Var}(X)
=
\frac{a^2+b^2+c^2-ab-ac-bc}{18}.
$$

Substituting \(a=4\), \(b=10\), and \(c=6\),

$$
\mathrm{Var}(X)
=
\frac{4^2+10^2+6^2-(4)(10)-(4)(6)-(10)(6)}{18}.
$$

Therefore,

$$
\mathrm{Var}(X)
=
\frac{16+100+36-40-24-60}{18}
=
\frac{28}{18}
=
\frac{14}{9}
\approx 1.56.
$$

---

## 6. Interpretation of Parameters

- \(a=4\): shortest possible installation time
- \(b=10\): longest possible installation time
- \(c=6\): most likely installation time

---

## 7. Why a Triangular Distribution?

A triangular distribution is useful when limited historical data is available, but experts can still estimate:

- minimum value,
- maximum value,
- most likely value.

It provides a simple bounded model for uncertainty.

---

## 8. Probability \(P(X \leq 5)\)

Using the first branch of the density:

$$
P(X\leq 5)
=
\int_{4}^{5} \frac{x-4}{6}\,dx.
$$

Thus,

$$
P(X\leq 5)
=
\frac{1}{6}\int_{4}^{5} (x-4)\,dx
=
\frac{1}{6}\cdot \frac{1}{2}
=
\frac{1}{12}.
$$

Hence,

$$
P(X\leq 5)
=
\frac{1}{12}
\approx 0.0833.
$$

---

## 9. Probability \(P(5 \leq X \leq 8)\)

Split the integral at \(x=6\):

$$
P(5\leq X\leq 8)
=
\int_{5}^{6} \frac{x-4}{6}\,dx
+
\int_{6}^{8} \frac{10-x}{12}\,dx.
$$

First part:

$$
\int_{5}^{6} \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_{5}^{6} (x-4)\,dx
=
\frac{1}{6}\cdot \frac{3}{2}
=
\frac{1}{4}.
$$

Second part:

$$
\int_{6}^{8} \frac{10-x}{12}\,dx
=
\frac{1}{12}\int_{6}^{8} (10-x)\,dx
=
\frac{1}{12}\cdot 6
=
\frac{1}{2}.
$$

Therefore,

$$
P(5\leq X\leq 8)
=
\frac{1}{4}+\frac{1}{2}
=
\frac{3}{4}.
$$

---

## 10. Practical Use

This model helps project planners:

- estimate completion probabilities,
- analyze scheduling uncertainty,
- plan staffing and contingency strategies,
- perform sensitivity analysis.
"""))

# ---------------------------------------------------
# Plot the Triangular Distribution
# ---------------------------------------------------

# Parameters
a = 4
b = 10
c_mode = 6

# Scipy parametrization
loc = a
scale = b - a
c_scipy = (c_mode - a) / (b - a)

# Generate x-values
x = np.linspace(a, b, 500)

# Compute PDF values
y = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)

# Create Plotly figure
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    fill='tozeroy',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=3)
))

# Add mode point
fig.add_trace(go.Scatter(
    x=[c_mode],
    y=[triang.pdf(c_mode, c=c_scipy, loc=loc, scale=scale)],
    mode='markers',
    marker=dict(size=10, color='red'),
    name='Mode (Most Likely Time)'
))

# Update layout
fig.update_layout(
    title='Triangular Distribution for Installation Time',
    xaxis_title='Installation Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white',
    width=900,
    height=500
)

# Display graph at bottom
fig.show()
```

```python



from IPython.display import display, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang, beta

# ---------------------------------------------------
# Display Mathematical Explanation
# ---------------------------------------------------

display(Markdown(r"""
# Comparison of Triangular and PERT Distributions

## 1. Given Parameters

The given parameters are

$$
a=8, \qquad b=20, \qquad c=11.
$$

The PERT shape parameters are

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
= 1 + 4\left( \frac{11-8}{20-8} \right)
= 2
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right)
= 1 + 4\left( \frac{20-11}{20-8} \right)
= 4.
$$

---

## 2. Support of the Random Variable

Since \(X\) represents the task duration in days, the support is

$$
8 \leq X \leq 20.
$$

---

## 3. Probability Density Function

The PERT density function is

$$
f_X(x)=
\frac{(x-8)^{\alpha-1}(20-x)^{\beta-1}}
{B(\alpha,\beta)(20-8)^{\alpha+\beta-1}},
\qquad 8 \leq x \leq 20.
$$

Substituting \(\alpha=2\) and \(\beta=4\),

$$
f_X(x)=
\frac{(x-8)(20-x)^3}
{B(2,4)\,12^5}.
$$

Using

$$
B(2,4)=\frac{\Gamma(2)\Gamma(4)}{\Gamma(6)}
=\frac{1!\,3!}{5!}
=\frac{1}{20},
$$

we obtain

$$
f_X(x)=
\frac{20(x-8)(20-x)^3}{12^5},
\qquad 8 \leq x \leq 20.
$$

and

$$
f_X(x)=0
\qquad \text{otherwise}.
$$

---

## 4. Probability Space

A suitable probability space is

$$
\Omega=[8,20], \qquad
\mathfrak F=\mathfrak B([8,20]),
$$

where \(\mathfrak B([8,20])\) is the Borel \(\sigma\)-algebra.

Define the probability measure by

$$
P(A)=\int_A f_X(x)\,dx,
\qquad A\in\mathfrak F.
$$

---

## 5. Random Variable Definition

Define

$$
X:\Omega \to \mathbb R
$$

by

$$
X(\omega)=\omega,
\qquad \omega\in[8,20].
$$

Thus \(X\) records the task duration in days.

---

## 6. Expectation of the PERT Distribution

The expectation formula becomes

$$
E[X]=\frac{a+4c+b}{6}.
$$

Substituting the values,

$$
E[X]
=
\frac{8+4(11)+20}{6}
=
\frac{72}{6}
=
12.
$$

Therefore,

$$
E[X]=12.
$$

---

## 7. Variance

The variance approximation is

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}.
$$

Substituting the values,

$$
\mathrm{Var}(X)
=
\frac{(12-8)(20-12)}{7}
=
\frac{32}{7}
\approx 4.57.
$$

---

## 8. Standard Deviation

$$
\sigma
=
\sqrt{\mathrm{Var}(X)}
=
\sqrt{\frac{32}{7}}
\approx 2.14.
$$

Using the rule of thumb,

$$
\sigma \approx \frac{b-a}{6}
=
\frac{20-8}{6}
=
2,
$$

which is close to the computed value.

---

## 9. Comparison Between Triangular and PERT Models

| Quantity | Triangular | PERT |
|---|---|---|
| Mean | \(13\) | \(12\) |
| Variance | \(\dfrac{49}{6}\approx 8.17\) | \(\dfrac{32}{7}\approx 4.57\) |
| Standard deviation | \(\approx 2.86\) | \(\approx 2.14\) |
| Main feature | Linear, sharp peak | Smooth, mode weighted more heavily |

The PERT model pulls the expected value closer to the most likely case and produces a smoother distribution with lower variance.

---

## 10. Interpretation

- \(a=8\): shortest plausible completion time
- \(b=20\): longest plausible completion time
- \(c=11\): most likely completion time
- \(E[X]=12\): expected completion time
- \(\sigma \approx 2.14\): uncertainty in task duration

These quantities are useful in project planning, deadline estimation, and risk assessment.
"""))

# ---------------------------------------------------
# Plotting Triangular vs PERT Distributions
# ---------------------------------------------------

# Parameters
a = 8
b = 20
c_mode = 11

# Triangular Distribution Parameters
c_scipy = (c_mode - a) / (b - a)
loc = a
scale = b - a

# PERT Distribution Parameters
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

# Generate x-values
x = np.linspace(a, b, 500)

# Compute PDFs
y_triang = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)
y_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b-a)

# Create figure
fig = go.Figure()

# Triangular Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_triang,
    mode='lines',
    name='Triangular Distribution',
    line=dict(color='royalblue', width=3, dash='dash'),
    fill='tozeroy'
))

# PERT Distribution
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    name='PERT Distribution',
    line=dict(color='firebrick', width=3)
))

# Add most likely value marker
fig.add_trace(go.Scatter(
    x=[c_mode],
    y=[max(y_pert)],
    mode='markers',
    marker=dict(size=10, color='black'),
    name='Most Likely Value'
))

# Layout settings
fig.update_layout(
    title='Comparison of Triangular and PERT Distributions',
    xaxis_title='Task Duration (days)',
    yaxis_title='Probability Density',
    template='plotly_white',
    width=950,
    height=550
)

# Display graph at bottom
fig.show()
```

```python
# Google Colab / Jupyter Notebook Display Code
# Symmetric PERT Distribution Example

from IPython.display import display, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

# ---------------------------------------------------
# Display Mathematical Explanation
# ---------------------------------------------------

display(Markdown(r"""
# Symmetric PERT Distribution

Use the PERT mean formula

$$
E[X]=\frac{a+4c+b}{6}.
$$

---

## 1. Condition for Symmetry

If the expected value equals the most likely value, then

$$
E[X]=c.
$$

Substitute into the PERT mean formula:

$$
\frac{a+4c+b}{6}=c.
$$

Multiply both sides by \(6\):

$$
a+4c+b=6c.
$$

Hence,

$$
a+b=2c.
$$

Solving for \(b\),

$$
b=2c-a.
$$

---

## 2. Numerical Example

Suppose

$$
a=5,
\qquad
c=11.
$$

Then

$$
b=2(11)-5
=22-5
=17.
$$

Thus,

$$
b=17.
$$

---

## 3. Variance Formula

Using the approximation

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7},
$$

and imposing the condition \(E[X]=c\),

$$
\mathrm{Var}(X)
=
\frac{(c-a)(b-c)}{7}.
$$

Since

$$
b=2c-a,
$$

we obtain

$$
b-c=(2c-a)-c=c-a.
$$

Therefore,

$$
\mathrm{Var}(X)
=
\frac{(c-a)^2}{7}.
$$

---

## 4. Equivalent Form

Because

$$
b-a=(2c-a)-a
=2(c-a),
$$

we have

$$
(c-a)^2
=
\frac{(b-a)^2}{4}.
$$

Substituting into the variance formula,

$$
\mathrm{Var}(X)
=
\frac{(b-a)^2}{28}.
$$

---

## 5. Final Result

If the expected value equals the most likely value, then the PERT distribution becomes symmetric and

$$
\boxed{
\mathrm{Var}_{\text{PERT}}(X)
=
\frac{(c-a)^2}{7}
=
\frac{(b-a)^2}{28}
}
$$

holds.

---

## 6. Interpretation

- \(a\): optimistic completion time
- \(b\): pessimistic completion time
- \(c\): most likely completion time

When \(E[X]=c\), the distribution is perfectly balanced around the mode, producing a symmetric PERT model.
"""))

# ---------------------------------------------------
# Plot Symmetric PERT Distribution
# ---------------------------------------------------

# Parameters
a = 5
c_mode = 11
b = 17

# PERT shape parameters
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

# Generate x-values
x = np.linspace(a, b, 500)

# Compute PDF
y_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b-a)

# Create Plotly figure
fig = go.Figure()

# Add PERT curve
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    fill='tozeroy',
    name='Symmetric PERT Distribution',
    line=dict(color='firebrick', width=3)
))

# Add mode / mean marker
fig.add_trace(go.Scatter(
    x=[c_mode],
    y=[max(y_pert)],
    mode='markers',
    marker=dict(size=11, color='black'),
    name='Mean = Mode'
))

# Layout settings
fig.update_layout(
    title='Symmetric PERT Distribution',
    xaxis_title='Task Duration',
    yaxis_title='Probability Density',
    template='plotly_white',
    width=900,
    height=520
)

# Display graph
fig.show()
```
